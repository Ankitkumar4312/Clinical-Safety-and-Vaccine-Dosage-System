```python
import pandas as pd
```

```python
df=pd.read_csv(r'D:\COMPUTER SCIENCE\PRIME BATCH\PROJECTS\6.Vaccine safety & Dosage\data\patient_records_cleaned.csv')
```

```python
df.head()
df.columns
```

```python
features=['Age', 'Biomarker_Level', 'Pre_Existing_Conditions', 'Vaccine_Dose_mcg',
       'Prior_Reaction_History']
```

```python
df.head(15)
```

```python
#standard scaler and split
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()

x=df.drop('Reaction_Score',axis=1)
y=df['Reaction_Score']
x_train,x_test,y_train,y_test=train_test_split(
    x,y,
    test_size=0.2,random_state=42
)

```

```python
#scaling of training data only
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
param_grids={
    'penalty':['l1','l2'],
    'C':[0.1,1,10,100],
    'solver':['liblinear','saga']
}
model=LogisticRegression(max_iter=100,class_weight='balanced',C=0.1)
grid_search=GridSearchCV(model,param_grids,cv=5,scoring='f1')
grid_search.fit(x_train_scaled,y_train)
```

```python
print("Best Parameters:", grid_search.best_params_)
print("Best F1 Score:", grid_search.best_score_)
```

```python
#best logistic regression
best_model=grid_search.best_estimator_
best_model.fit(x_train_scaled,y_train)
```

```python
y_pred=best_model.predict(x_test_scaled)
```

```python
#comparision
comparision=pd.DataFrame({
    'Actual':y_test.values,
    'predicted':y_pred
})
print(comparision.head(15))
```

```python
#evaluation metrics - LOGISTIC REGRESSION
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score,confusion_matrix,classification_report

print("Accuracy_score:",accuracy_score(y_pred,y_test))
print("recall_score:",recall_score(y_pred,y_test))
print("precision_score:",precision_score(y_pred,y_test))
print("f1_score:",f1_score(y_pred,y_test))
print("confusion_matrix:\n",confusion_matrix(y_pred,y_test))
print("Classification_report:\n",classification_report(y_pred,y_test))
```

```python
#lets save this model 
#let save with joblib
import joblib
joblib.dump(best_model,"Logistic_Regression.pkl")
joblib.dump(scaler,"Scaler.pkl")
print("Logistic_Regression Saved Successfully!!!")
```
