```python
import pandas as pd
print('Libraries imported successfully!!!')
```

```python
#Loading of Data
df=pd.read_csv('patient_records.csv')
print('Number of Rows and columns :',df.shape)
```

```python
#information related to Data
df.info()
df.describe()
```

```python
#checking for duplicated data
print('duplicted data :',df.duplicated().sum())
```

```python
#checking of missing values 
print('Missing values:\n',df.isnull().sum())
```

```python
#removing the patient id column as it does not effect the result
df=df.drop('Patient_ID',axis=1)
```

```python
#Handling missing values
df['Biomarker_Level']=df['Biomarker_Level'].fillna(5.28)
df.isnull().sum()
df['Prior_Reaction_History']=df['Prior_Reaction_History'].fillna('None')
print('Missing values successully resolved!!!')
```

```python
df.head(15)
```

```python
#encoding or mapping the Prior_Reaction_History column
df['Prior_Reaction_History']=df['Prior_Reaction_History'].map({
    'None':0,
    'Mild':1,
    'Moderate':2
})
```
